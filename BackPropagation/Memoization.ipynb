# Memoization in Deep Learning

## What is Memoization?

**Memoization** is an optimization technique in which the results of previously computed values are **stored (cached)** so that they do not have to be calculated again.

Instead of performing the same computation multiple times, the stored result is reused whenever needed.

> **In simple words:**  
> Memoization means **"Compute once, store it, and reuse it whenever required."**

---

# Real-Life Analogy

Imagine you are solving a difficult math problem.

Without memoization:

```
Need answer

↓

Solve again

↓

Need answer again

↓

Solve again

↓

Need answer again

↓

Solve again
```

This wastes time.

With memoization:

```
Solve once

↓

Write the answer in your notebook

↓

Next time

↓

Read from notebook

↓

Done
```

This saves both time and computation.

---

# Why is Memoization Important?

Without memoization,

- Same calculations are repeated.
- Training becomes slower.
- More computational power is required.

With memoization,

- Faster computation
- Less repeated work
- Efficient backpropagation
- Lower computational cost

---

# Memoization in Machine Learning

In Machine Learning, memoization is used whenever intermediate values are needed multiple times.

Examples:

- Dynamic Programming
- Neural Networks
- Graph Algorithms
- Recursive Algorithms

---

# Memoization in Multi-Layer Perceptron (MLP)

## What is an MLP?

A **Multi-Layer Perceptron (MLP)** is a feedforward neural network consisting of:

- Input Layer
- One or more Hidden Layers
- Output Layer

Example:

```
Input

↓

Hidden Layer 1

↓

Hidden Layer 2

↓

Output
```

---

## Why is Memoization Needed in MLP?

During training, an MLP has two phases:

### 1. Forward Propagation

- Inputs pass through the network.
- Every neuron computes:
  - Weighted Sum (Z)
  - Activation (A)

Example:

```
Input

↓

Z1

↓

A1

↓

Z2

↓

A2

↓

Prediction
```

---

### 2. Backpropagation

During backpropagation, we need:

- Z values
- Activation values
- Outputs of every layer

If these values were not stored during forward propagation,

the network would have to recompute everything again.

This would be very slow.

Therefore,

**MLPs store all intermediate values during forward propagation.**

This storage is called **memoization (or caching)**.

---

# Example

Forward Pass

```
Input

↓

Z1 = W1X + b1

↓

A1 = ReLU(Z1)

↓

Z2 = W2A1 + b2

↓

A2 = Sigmoid(Z2)

↓

Output
```

During Forward Pass, we store:

```
Z1

A1

Z2

A2
```

These stored values are later reused during backpropagation.

---

# What Values are Stored?

During Forward Propagation, the following are commonly stored:

- Input values
- Weighted sums (Z)
- Activation outputs (A)
- Predicted output
- Loss

These cached values are reused while computing gradients.

---

# Advantages of Memoization in MLP

- Faster backpropagation
- No repeated calculations
- Saves computation time
- Improves training efficiency
- Essential for deep neural networks

---

# Disadvantages

- Requires additional memory
- Very deep networks store many intermediate values
- Higher RAM/GPU memory usage

---

# Backpropagation = Chain Rule + Memoization

One of the most important ideas in Deep Learning is:

```
Backpropagation
=
Chain Rule
+
Memoization
```

---

# What is Backpropagation?

Backpropagation is the algorithm used to train neural networks.

Its main purpose is to:

- Calculate gradients
- Update weights
- Reduce the loss

---

# Two Main Components of Backpropagation

## 1. Chain Rule (Differentiation)

The **Chain Rule** is a rule from calculus used to compute derivatives of composite functions.

Since a neural network is made of many layers,

the output depends on the previous layer,

which depends on another previous layer,

and so on.

Therefore, derivatives are calculated using the Chain Rule.

---

## Example

Suppose

```
Loss

↓

Output

↓

Hidden Layer 2

↓

Hidden Layer 1

↓

Input
```

The gradient flows backward through each layer using the Chain Rule.

Mathematically,

$$
\frac{\partial L}{\partial W}
=
\frac{\partial L}{\partial A}
\times
\frac{\partial A}{\partial Z}
\times
\frac{\partial Z}{\partial W}
$$

This is the Chain Rule in action.

---

# 2. Memoization (Caching)

While applying the Chain Rule,

many intermediate values are needed repeatedly.

Examples:

- Z values
- Activation outputs
- Previous layer outputs

Instead of recomputing them,

the forward pass stores them in memory.

Backpropagation simply reuses them.

This reuse is called **Memoization**.

---

# How Backpropagation Works

### Step 1

Forward Propagation

```
Input

↓

Compute Z

↓

Compute Activation

↓

Store Z and A

↓

Prediction
```

---

### Step 2

Calculate Loss

```
Prediction

↓

Loss
```

---

### Step 3

Backward Propagation

```
Loss

↓

Use Stored Values

↓

Apply Chain Rule

↓

Calculate Gradients

↓

Update Weights
```

Notice that the stored values from the forward pass are reused instead of being computed again.

---

# Why Backpropagation Needs Memoization

Without memoization,

for every weight update,

the network would need to calculate:

```
Z1

↓

A1

↓

Z2

↓

A2

↓

...

Again
```

This would make training extremely slow.

With memoization,

```
Forward Pass

↓

Store Z and A

↓

Backpropagation

↓

Reuse Stored Values
```

Training becomes much faster.

---

# Chain Rule vs Memoization

| Chain Rule | Memoization |
|------------|-------------|
| Mathematical concept | Programming optimization technique |
| Computes gradients | Stores intermediate values |
| Uses differentiation | Uses caching |
| Required for gradient calculation | Required for efficient computation |
| Comes from calculus | Comes from computer science |

---

# Backpropagation Workflow

```
Input
   ↓
Forward Propagation
   ↓
Compute Z
   ↓
Compute A
   ↓
Store (Memoization)
   ↓
Prediction
   ↓
Loss
   ↓
Backward Propagation
   ↓
Use Chain Rule
   ↓
Reuse Stored Values
   ↓
Compute Gradients
   ↓
Update Weights
```

---

# Key Points to Remember

- **Memoization** means storing previously computed values and reusing them instead of recomputing them.
- In an **MLP**, intermediate values such as **weighted sums (Z)** and **activation outputs (A)** are stored during the forward pass.
- During **backpropagation**, these stored values are reused to compute gradients efficiently.
- **Backpropagation** is based on two ideas:
  1. **Chain Rule** – Computes gradients through multiple layers.
  2. **Memoization** – Stores and reuses intermediate values to avoid repeated computations.
- Memoization speeds up training but requires additional memory to store intermediate results.
