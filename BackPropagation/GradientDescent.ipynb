# Gradient Descent in Neural Networks

## What is Gradient Descent?

**Gradient Descent** is an optimization algorithm used to train Machine Learning and Deep Learning models.

Its main goal is to **minimize the loss function** by updating the weights and biases of the neural network.

> **In simple words:**  
> Gradient Descent helps the neural network learn by gradually moving the weights in the direction that reduces the prediction error.

---

# Why Do We Need Gradient Descent?

Initially, the weights of a neural network are assigned randomly.

Random weights lead to poor predictions.

Gradient Descent repeatedly updates these weights until the loss becomes as small as possible.

```
Random Weights
      ↓
Prediction
      ↓
Calculate Loss
      ↓
Calculate Gradient
      ↓
Update Weights
      ↓
Better Prediction
```

---

# Real-Life Analogy

Imagine you are standing on top of a mountain and want to reach the lowest point (valley).

Since it is dark, you cannot see the entire mountain.

You take one small step in the direction where the ground slopes downward.

You continue taking small downhill steps until you reach the bottom.

Similarly,

- Mountain = Loss Function
- Bottom = Minimum Loss
- Step = Weight Update
- Gradient = Direction of steepest increase
- Gradient Descent = Move opposite to the gradient

---

# Working of Gradient Descent

Step 1

Forward Propagation

```
Input

↓

Prediction
```

---

Step 2

Calculate Loss

```
Prediction

↓

Loss
```

---

Step 3

Backpropagation

```
Loss

↓

Gradient
```

---

Step 4

Update Weights

The weights are updated using:

## Weight Update Formula

$$
W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}
$$

### Where,

- \(W_{old}\) = Current weight
- \(W_{new}\) = Updated weight
- \(\eta\) = Learning Rate
- \(L\) = Loss Function
- \(\frac{\partial L}{\partial W}\) = Gradient of Loss with respect to Weight

---

# Learning Rate (η)

The **Learning Rate** determines the size of each step during weight updates.

### Small Learning Rate

```
Tiny Steps

↓

Slow Learning

↓

May take many epochs
```

---

### Large Learning Rate

```
Very Big Steps

↓

May overshoot minimum

↓

Training becomes unstable
```

---

### Good Learning Rate

```
Balanced Steps

↓

Fast and Stable Learning
```

# Types of Gradient Descent

There are three main types:

1. Batch Gradient Descent (BGD)
2. Stochastic Gradient Descent (SGD)
3. Mini-Batch Gradient Descent (MBGD)

---

# 1. Batch Gradient Descent (BGD)

## Definition

Batch Gradient Descent uses the **entire training dataset** to calculate the gradient before updating the weights.

Example

Dataset Size

```
10000 samples
```

BGD uses

```
All 10000 samples

↓

Compute Gradient

↓

Update Weights Once
```

---

## Working

```
Entire Dataset

↓

Forward Pass

↓

Loss

↓

Gradient

↓

Weight Update
```

---

## Advantages

- Stable gradient updates
- Smooth convergence
- Good for small datasets

---

## Disadvantages

- Very slow for large datasets
- High memory usage
- One update only after processing the entire dataset

---

# 2. Stochastic Gradient Descent (SGD)

## Definition

Stochastic Gradient Descent updates the weights **after every single training example**.

Example

Dataset

```
10000 samples
```

Processing

```
Sample 1

↓

Update

↓

Sample 2

↓

Update

↓

Sample 3

↓

Update

...
```

---

## Working

```
One Sample

↓

Forward Pass

↓

Loss

↓

Gradient

↓

Update
```

---

## Advantages

- Very fast updates
- Low memory usage
- Suitable for very large datasets
- Can escape local minima because of noisy updates

---

## Disadvantages

- Noisy updates
- Loss fluctuates
- Less stable convergence

---

# 3. Mini-Batch Gradient Descent (MBGD)

## Definition

Mini-Batch Gradient Descent divides the dataset into **small batches**.

Each batch is used to calculate gradients and update weights.

Example

Dataset

```
10000 samples
```

Batch Size

```
100
```

Processing

```
Batch 1 (100)

↓

Update

↓

Batch 2 (100)

↓

Update

↓

Batch 3 (100)

↓

Update
```

---

## Advantages

- Faster than BGD
- More stable than SGD
- Efficient GPU utilization
- Most commonly used in Deep Learning

---

## Disadvantages

- Need to choose batch size
- Slightly more memory than SGD

---

# Comparison of BGD, SGD and MBGD

| Feature | Batch GD | SGD | Mini-Batch GD |
|----------|----------|-----|---------------|
| Data Used | Entire Dataset | One Sample | Small Batch |
| Weight Updates | Once per Epoch | After Every Sample | After Every Batch |
| Memory Usage | High | Low | Medium |
| Speed | Slow | Fast | Fast |
| Stability | High | Low | High |
| GPU Efficiency | Poor | Poor | Excellent |
| Used in Deep Learning | Rarely | Sometimes | Almost Always |

---

# Which is Faster to Converge?

**Convergence** means reaching the minimum loss.

| Algorithm | Convergence |
|-----------|-------------|
| Batch GD | Slow |
| SGD | Fast but noisy |
| Mini-Batch GD | Fast and stable (Best) |

### Answer

**Mini-Batch Gradient Descent converges the fastest in practice** because it combines the speed of SGD with the stability of Batch Gradient Descent.

---

# Which is Faster?

Here, "faster" refers to the **time taken for one weight update**.

| Algorithm | Speed per Update |
|-----------|------------------|
| SGD | Fastest |
| Mini-Batch GD | Fast |
| Batch GD | Slowest |

### Answer

- **Fastest per update:** SGD
- **Best overall training performance:** Mini-Batch Gradient Descent

# Vectorization

## What is Vectorization?

**Vectorization** is the process of performing computations on an **entire vector or matrix at once** instead of using loops.

Modern CPUs and GPUs are highly optimized for matrix operations.

---

## Without Vectorization

```
for sample 1

↓

Calculate

↓

for sample 2

↓

Calculate

↓

for sample 3

↓

Calculate
```

Many repeated operations.

---

## With Vectorization

```
Entire Batch

↓

Single Matrix Multiplication

↓

Output
```

No explicit loops are required.

---

# Why is Vectorization Faster?

Because matrix operations are:

- Parallel
- GPU optimized
- Memory efficient

Example

Without Vectorization

```
1000 calculations

↓

One by One
```

With Vectorization

```
1000 calculations

↓

Single Matrix Operation
```

---

# Advantages of Vectorization

- Faster execution
- Better GPU utilization
- Less code
- Lower computational overhead

---

# Why is Batch Size Usually a Multiple of 2?

Common batch sizes are:

```
16

32

64

128

256

512
```

---

## Reason 1: Better Memory Alignment

Computers store data in binary.

Powers of 2 fit naturally into memory blocks.

This improves memory access speed.

---

## Reason 2: GPU Optimization

Modern GPUs are designed to process data most efficiently when batch sizes are powers of 2.

This increases parallel processing efficiency.

---

## Reason 3: Faster Matrix Operations

Deep learning relies heavily on matrix multiplication.

Batch sizes such as:

```
32

64

128
```

allow optimized matrix multiplication algorithms.

---

## Is It Mandatory?

**No.**

Batch size **does not have to be a power of 2**.

You can use:

```
30

50

75

100
```

The model will still train correctly.

Using powers of 2 is simply an optimization for hardware performance.

---

# What if Batch Size Does Not Divide the Number of Rows?

Suppose:

```
Dataset = 1030 samples

Batch Size = 128
```

Batches become

```
128

128

128

128

128

128

128

128

6
```

The final batch contains only **6 samples**.

---

## Is This a Problem?

**No.**

Modern deep learning frameworks (TensorFlow, PyTorch, Keras) automatically process the last smaller batch.

Training continues normally.

---

## Example

Dataset

```
1000 samples
```

Batch Size

```
64
```

Number of batches

```
64

64

64

...

40
```

The last batch contains **40 samples**.

---

## What Does `drop_last=True` Mean?

Some frameworks provide an option called:

```
drop_last=True
```

If enabled,

the incomplete last batch is discarded.

Example

Dataset

```
1000 samples
```

Batch Size

```
64
```

Normally

```
15 batches of 64

+

1 batch of 40
```

With

```
drop_last=True
```

Only

```
15 batches of 64
```

are used.

The last 40 samples are ignored for that epoch.

# Complete Gradient Descent Workflow

```
Training Data
      ↓
Forward Propagation
      ↓
Prediction
      ↓
Loss Function
      ↓
Backpropagation
      ↓
Gradient Calculation
      ↓
Gradient Descent
      ↓
Weight Update
      ↓
Repeat Until Minimum Loss
```

---

# Key Points to Remember

- **Gradient Descent** minimizes the loss function by updating weights in the opposite direction of the gradient.
- The weight update rule is:

  $$
  W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}
  $$

- **Batch Gradient Descent (BGD)** uses the entire dataset for one update. It is stable but slow.
- **Stochastic Gradient Descent (SGD)** updates weights after every training sample. It is the fastest per update but has noisy convergence.
- **Mini-Batch Gradient Descent (MBGD)** updates weights after processing a small batch of samples. It provides the best balance between speed and stability and is the most widely used method in Deep Learning.
- **Vectorization** speeds up training by performing matrix operations on entire batches instead of processing samples one by one.
- Batch sizes are commonly chosen as **powers of 2 (16, 32, 64, 128, ...)** because they make better use of GPU hardware and memory, but this is **not mandatory**.
- If the dataset size is **not divisible by the batch size**, the final batch simply contains fewer samples. Most deep learning frameworks handle this automatically, or the incomplete batch can be discarded using options like `drop_last=True`.
