# <span style="color:red">Core Multi-Layer Perceptron (MLP) Notation</span>

To scale from a single perceptron to a **Multi-Layer Perceptron (MLP)**, we track numbers across different layers, rows, and connections using strict index notation. 



---

## <span style="color:blue">1. The Structural Indexes</span>

We use three primary letter trackers to locate any variable inside the network:
* $l$: The **current layer** we are looking at (where $l=0$ is the input layer).
* $j$: The **target neuron** in the current layer ($l$).
* $k$: The **source neuron** in the previous layer ($l-1$).

---

## <span style="color:blue">2. Core Variable Definitions</span>

### Weights ($W$)
$$w_{jk}^{[l]}$$
* **Meaning:** The weight connecting neuron $k$ (in the previous layer $l-1$) to neuron $j$ (in the current layer $l$).
* *Order Note:* The target neuron index ($j$) comes *before* the source neuron index ($k$) to make matrix multiplication math align perfectly.

### Biases ($b$)
$$b_{j}^{[l]}$$
* **Meaning:** The bias value added to neuron $j$ in layer $l$.

### Linear Combination ($z$)
$$z_{j}^{[l]}$$
* **Meaning:** The raw, pre-activation sum computed inside neuron $j$ in layer $l$.
* **Formula:** $$z_{j}^{[l]} = \sum_{k} \left( w_{jk}^{[l]} \cdot a_{k}^{[l-1]} \right) + b_{j}^{[l]}$$

### Activation Output ($a$)
$$a_{j}^{[l]}$$
* **Meaning:** The final output of neuron $j$ in layer $l$ after running through the activation function $\sigma$ (like ReLU or Sigmoid).
* **Formula:** $$a_{j}^{[l]} = \sigma(z_{j}^{[l]})$$
* *Special Case:* $a^{[0]} = x$ (The activation of the 0th layer is just your raw input features).

---

## <span style="color:blue">3. Compact Matrix Notation</span>

When writing code (like with NumPy or PyTorch), we process entire layers at once using matrices instead of looping through single neurons:

$$Z^{[l]} = W^{[l]} \cdot A^{[l-1]} + b^{[l]}$$
$$A^{[l]} = \sigma(Z^{[l]})$$

---

## <span style="color:blue">Quick Variable Cheat Sheet</span>

| Variable Notation | Core Definition | Dimension Context |
| :--- | :--- | :--- |
| **$l$** | Layer tracking index | $l = 1, 2, \dots, L$ |
| **$W^{[l]}$** | Weight matrix for layer $l$ | $(\text{Neurons in } l \times \text{Neurons in } l-1)$ |
| **$b^{[l]}$** | Bias vector for layer $l$ | $(\text{Neurons in } l \times 1)$ |
| **$Z^{[l]}$** | Pre-activation vector for layer $l$ | $(\text{Neurons in } l \times 1)$ |
| **$A^{[l]}$** | Activated output vector for layer $l$| $(\text{Neurons in } l \times 1)$ |
