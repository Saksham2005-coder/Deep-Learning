# <span style="color:red">Forward Propagation Mechanics</span>

In a Multi-Layer Perceptron (MLP), **Forward Propagation** is the process where input data moves forward through the layers of the network to generate a final prediction ($\hat{y}$).

---

## <span style="color:pink">1. The Two-Step Computational Engine</span>

Every single hidden and output neuron in a neural network operates like an assembly line, executing two steps in order:

### Step A: The Linear Step ($Z$)
The neuron pulls all incoming outputs from the previous layer, multiplies each by its corresponding weight ($W$), sums them up, and adds a local bias ($b$).
$$z = w_1a_1 + w_2a_2 + \dots + w_na_n + b$$

### Step B: The Non-Linear Activation Step ($A$)
The raw score $z$ is squished through an activation function (like **ReLU** or **Sigmoid**). This introduces non-linearity, allowing the network to learn curved patterns instead of just straight lines.
$$a = \sigma(z)$$

---

## <span style="color:pink">2. Vectorized Layer Equations</span>

To process an entire layer of neurons simultaneously, we use matrix mathematics instead of looping through individual nodes:

$$Z^{[l]} = W^{[l]} \cdot A^{[l-1]} + b^{[l]}$$
$$A^{[l]} = \sigma(Z^{[l]})$$

### Variable Guide:
* **$A^{[l-1]}$**: The matrix of outputs coming from the *previous* layer.
* **$W^{[l]}$**: The weight matrix of the *current* layer.
* **$b^{[l]}$**: The bias vector of the *current* layer.
* **$Z^{[l]}$**: The pre-activation matrix (raw results) of the *current* layer.
* **$A^{[l]}$**: The final activated output matrix of the *current* layer.

---

## <span style="color:pink">3. Full Network Workflow Matrix</span>

| Execution Stage | Layer Designation | Mathematical Equation | Purpose |
| :--- | :--- | :--- | :--- |
| **0. Initialization** | Input Layer ($l=0$) | $A^{[0]} = X$ | Loads raw dataset features into the system. |
| **1. Hidden Processing**| Hidden Layer ($l=1$) | $Z^{[1]} = W^{[1]}A^{[0]} + b^{[1]}$<br>$A^{[1]} = \text{ReLU}(Z^{[1]})$ | Extracts low-level interaction features. |
| **2. Final Target** | Output Layer ($l=2$) | $Z^{[2]} = W^{[2]}A^{[1]} + b^{[2]}$<br>$\hat{y} = \text{Sigmoid}(Z^{[2]})$ | Computes the final probability score. |
