# Data Scaling in Neural Networks

## What is Data Scaling?

**Data Scaling** (or **Feature Scaling**) is the process of transforming numerical features so that they have a similar range or distribution.

In Neural Networks, feature scaling is one of the most important preprocessing steps because neural networks learn much faster when all input features are on a similar scale.

> **In simple words:**  
> Data Scaling means **bringing all feature values to a similar range** so that no feature dominates the learning process.

---

# Why Do We Need Data Scaling?

Consider a dataset containing:

| Feature | Value |
|----------|--------|
| Age | 25 |
| Salary | 8,50,000 |

Here,

```
Age → 25

Salary → 850000
```

The salary values are much larger than age values.

During training,

the neural network gives more importance to larger-valued features, even if they are not more important.

This results in:

- Slow training
- Poor convergence
- Unstable gradient updates

Feature Scaling solves this problem.

---

# Example

## Before Scaling

| Age | Salary |
|------|---------|
|22|300000|
|35|800000|
|45|1500000|

Notice that Salary is much larger than Age.

---

## After Scaling

| Age | Salary |
|------|---------|
|0.15|0.20|
|0.55|0.60|
|0.85|0.95|

Now both features have similar ranges.

---

# Why is Feature Scaling Important in ANN?

Neural Networks use:

- Gradient Descent
- Backpropagation
- Matrix Multiplication

Large feature values can produce:

- Large gradients
- Unstable weight updates
- Slow convergence

Scaling helps by:

- Faster convergence
- Stable gradient descent
- Better optimization
- Improved model performance

---

# Benefits of Feature Scaling

- Faster training
- Better convergence
- Stable gradient updates
- Prevents one feature from dominating
- Improves overall accuracy

---

# Feature Scaling in Artificial Neural Networks (ANN)

Feature Scaling is almost always recommended before training an ANN.

This is because:

- Weights are initialized with small values.
- Large feature values produce very large activations.
- Very small feature values produce tiny activations.
- Scaling keeps activations balanced.

---

# Types of Feature Scaling

There are two most common methods:

1. Normalization (Min-Max Scaling)
2. Standardization (Z-Score Scaling)

```
Feature Scaling
      │
      ├── Normalization
      └── Standardization
```

---

# 1. Normalization (Min-Max Scaling)

## What is Normalization?

Normalization scales all feature values to a fixed range, usually:

```
0 to 1
```

or sometimes

```
-1 to 1
```

The relative order of the data remains the same.

---

## Formula

$$
X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}
$$

### Where,

- \(X\) = Original value
- \(X_{min}\) = Minimum value of the feature
- \(X_{max}\) = Maximum value of the feature
- \(X_{scaled}\) = Normalized value

---

# Example

Original Data

| Marks |
|--------|
|20|
|40|
|60|
|80|
|100|

Here,

```
Minimum = 20

Maximum = 100
```

Normalize

```
20 → 0

40 → 0.25

60 → 0.50

80 → 0.75

100 → 1
```

---

# Characteristics

- Output range is fixed.
- Easy to interpret.
- Sensitive to outliers.

---

# Advantages

- Very easy to understand.
- Keeps values within a fixed range.
- Speeds up gradient descent.
- Works well when data has known minimum and maximum values.

---

# Disadvantages

- Highly affected by outliers.
- New maximum or minimum values require rescaling.

---

# Applications

- Image Processing
- Computer Vision
- Deep Learning
- CNNs
- Pixel Value Scaling

---

# 2. Standardization (Z-Score Scaling)

## What is Standardization?

Standardization transforms the data so that:

- Mean becomes **0**
- Standard Deviation becomes **1**

Unlike normalization, the values are **not restricted to 0–1**.

---

## Formula

$$
Z=\frac{X-\mu}{\sigma}
$$

### Where,

- \(X\) = Original value
- \(\mu\) = Mean of the feature
- \(\sigma\) = Standard Deviation
- \(Z\) = Standardized value

---

# Example

Original Data

```
20

30

40

50

60
```

After Standardization

```
-1.41

-0.71

0

0.71

1.41
```

Notice:

Mean

```
0
```

Standard Deviation

```
1
```

---

# Characteristics

- Mean becomes zero.
- Standard deviation becomes one.
- Values may be positive or negative.
- Less affected by outliers than normalization.

---

# Advantages

- Handles large ranges effectively.
- Less sensitive to outliers.
- Suitable for many Machine Learning and Deep Learning algorithms.
- Stable gradient updates.

---

# Disadvantages

- Output is not bounded.
- Slightly harder to interpret.

---

# Applications

- Artificial Neural Networks
- Logistic Regression
- Linear Regression
- Support Vector Machines (SVM)
- PCA

---

# Normalization vs Standardization

| Feature | Normalization | Standardization |
|----------|---------------|-----------------|
| Range | Usually 0 to 1 | No fixed range |
| Formula | \((X-X_{min})/(X_{max}-X_{min})\) | \((X-\mu)/\sigma\) |
| Mean | Not fixed | 0 |
| Standard Deviation | Not fixed | 1 |
| Outlier Sensitivity | High | Lower |
| Best For | Image Data, CNNs | ANN, Regression, SVM |

---

# Which Scaling Method Should You Use?

| Situation | Recommended Method |
|------------|-------------------|
| Neural Networks (ANN) | Standardization (Most Common) |
| CNN Image Pixels | Normalization |
| Logistic Regression | Standardization |
| Linear Regression | Standardization |
| SVM | Standardization |
| Features already in a fixed range | Normalization |

---

Without scaling,

```
Feature 1

Age

20

25

30

Feature 2

Salary

900000

1200000

1500000
```

During Gradient Descent,

Salary dominates the learning.

# Key Points to Remember

- **Feature Scaling** transforms numerical features to a similar scale before training.
- It is highly recommended for **Artificial Neural Networks (ANNs)** because it improves the performance of Gradient Descent and Backpropagation.
- **Normalization (Min-Max Scaling)** scales data to a fixed range, usually **0 to 1**, using:

  $$
  X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}
  $$

- **Standardization (Z-Score Scaling)** transforms data to have **mean = 0** and **standard deviation = 1**, using:

  $$
  Z=\frac{X-\mu}{\sigma}
  $$

- **Normalization** is commonly used for image pixel values and when the data has a known minimum and maximum.
- **Standardization** is generally preferred for **ANNs**, Logistic Regression, Linear Regression, SVM, and many other machine learning algorithms because it provides more stable optimization.
- Feature scaling leads to **faster convergence, stable weight updates, and better model performance**.

# Feature Scaling and Data Transformation (Simplified)

## 1. Normalization (Min-Max Scaling)
- Used to scale values into a fixed range (usually **0 to 1**).
- Done using **MinMaxScaler** in Scikit-Learn.
- You can change the range (e.g., **-1 to 1**) using the `feature_range` parameter.
- **Problem:** Very sensitive to outliers.

**Example:**
- Original: 10, 20, 30
- Scaled: 0, 0.5, 1

---

## 2. Standardization
- Converts data so that:
  - Mean = **0**
  - Standard Deviation = **1**
- Done using **StandardScaler**.
- Does **not** limit values to a fixed range.
- **Better than normalization when outliers exist.**

---

## 3. Data with a Heavy Tail
A heavy-tailed distribution has a few very large (or very small) values.

### Problem
- Min-Max Scaling and Standardization compress most values into a small range.
- This can reduce model performance.

### Solutions
- Apply a transformation before scaling:
  - **Square Root (`√x`)** → for moderately skewed positive data.
  - **Power Transformation (`x^p`, where 0 < p < 1)** → reduces skewness.
  - **Log Transformation (`log(x)`)** → for very heavily skewed data.
- Another option is **Bucketization (Binning)**:
  - Divide values into groups (buckets).
  - Replace each value with its bucket number.

---

## 4. Data with Multiple Peaks (Multimodal Data)
Multimodal data has **two or more peaks (modes)**.

### Solutions

### A. Bucketization + One-Hot Encoding
- Divide data into buckets.
- Treat bucket numbers as **categories**, not numbers.
- Encode them using **OneHotEncoder**.
- Helps the model learn different rules for different value ranges.

### B. Radial Basis Function (RBF)
- Create a new feature for each important peak.
- Measure how close a value is to each peak.
- The most common RBF is the **Gaussian RBF**, whose value decreases as the distance from the peak increases.

---

# Summary Table

| Data Type | Recommended Method |
|------------|--------------------|
| Normal data | Standardization |
| Data without outliers | Normalization (Min-Max Scaling) |
| Heavy-tailed data | Log, Square Root, Power Transform, then Scale |
| Multimodal data | Bucketization + One-Hot Encoding or Gaussian RBF |
