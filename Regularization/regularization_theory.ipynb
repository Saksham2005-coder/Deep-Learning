# Regularization in Deep Learning

## What is Regularization?

**Regularization** is a technique used in Deep Learning to **reduce overfitting** and improve the model's ability to generalize to unseen data.

Instead of allowing the neural network to memorize the training data, regularization forces it to learn simpler and more meaningful patterns.

> **In simple words:**  
> Regularization tells the neural network:
>
> **"Do not make the model unnecessarily complex. Learn only the important patterns."**

---

# Why Do We Need Regularization?

Suppose a neural network gives:

Training Accuracy

```
99%
```

Testing Accuracy

```
78%
```

The model has memorized the training data.

This is called **Overfitting**.

Regularization helps reduce this problem.

---

# How Regularization Works

Without Regularization

```
Complex Model

↓

Memorizes Training Data

↓

Overfitting
```

With Regularization

```
Simpler Model

↓

Learns General Patterns

↓

Better Generalization
```

---

# Types of Regularization in Deep Learning

There are several regularization techniques.

```
Regularization
      │
      ├── L1 Regularization (Lasso)
      ├── L2 Regularization (Ridge)
      ├── Dropout
      ├── Early Stopping
      ├── Data Augmentation
      └── Batch Normalization
```

In this section, we focus on **L1**, **L2**, and **Weight Decay**.

---

# L2 Regularization in ANN (Ridge Regularization)

## What is L2 Regularization?

**L2 Regularization** adds a penalty proportional to the **square of the weights** to the loss function.

This discourages the network from assigning very large values to weights.

Instead of minimizing only the prediction error,

the model minimizes:

```
Prediction Error

+

Penalty for Large Weights
```

---

# Formula

## Original Loss

$$
Loss
$$

---

## Loss with L2 Regularization

$$
Loss_{new}=Loss+\lambda\sum_{i=1}^{n}w_i^2
$$

### Where,

- \(Loss\) = Original loss function
- \(\lambda\) = Regularization parameter
- \(w_i\) = Weight
- \(n\) = Total number of weights

---

# How L2 Regularization Works

Suppose

```
Weights

20

18

25

15
```

Large weights make the model very complex.

After L2 Regularization

```
Weights

4

5

6

3
```

The weights become smaller.

The model becomes simpler.

---

# Effect of Lambda (λ)

### λ = 0

```
No Regularization
```

Model behaves normally.

---

### Small λ

```
Light Regularization
```

Slight reduction in overfitting.

---

### Large λ

```
Strong Regularization
```

Weights become very small.

Too large a value may lead to **underfitting**.

---

# Advantages of L2 Regularization

- Reduces overfitting
- Keeps weights small
- Improves generalization
- Stable training
- Works well with neural networks

---

# Disadvantages of L2 Regularization

- Does not remove unnecessary features.
- Requires choosing a suitable value of λ.
- Large λ may cause underfitting.

---

# L1 Regularization (Lasso Regularization)

## What is L1 Regularization?

**L1 Regularization** adds a penalty proportional to the **absolute value of the weights**.

Unlike L2,

L1 can reduce some weights **exactly to zero**.

This automatically removes less important features.

---

# Formula

$$
Loss_{new}=Loss+\lambda\sum_{i=1}^{n}|w_i|
$$

### Where,

- \(Loss\) = Original loss function
- \(\lambda\) = Regularization parameter
- \(w_i\) = Weight

---

# How L1 Regularization Works

Suppose

Original Weights

```
15

8

0.5

0.02

12
```

After L1

```
14

7

0

0

11
```

Some weights become exactly zero.

These neurons/features contribute very little to the prediction.

---

# Feature Selection

Since L1 makes some weights exactly zero,

those corresponding features are effectively ignored.

Therefore,

L1 performs **automatic feature selection**.

---

# Advantages of L1 Regularization

- Reduces overfitting
- Performs feature selection
- Produces sparse models
- Removes unimportant features automatically

---

# Disadvantages of L1 Regularization

- Training may be less stable than L2.
- May remove useful features if λ is too large.
- Optimization is slightly more difficult.

---

# L1 vs L2 Regularization

| Feature | L1 Regularization | L2 Regularization |
|----------|------------------|-------------------|
| Penalty | Absolute weights | Squared weights |
| Formula | \(\lambda\sum|w|\) | \(\lambda\sum w^2\) |
| Large Weights | Reduced | Reduced |
| Small Weights | Become Zero | Become Smaller |
| Feature Selection | Yes | No |
| Sparse Model | Yes | No |
| Stability | Moderate | High |

---

# Weight Decay in ANN

## What is Weight Decay?

**Weight Decay** is a technique that **reduces the weights slightly after every update**.

Instead of allowing weights to grow continuously,

Weight Decay shrinks them during training.

> **In simple words:**  
> Weight Decay continuously tells the weights:
>
> **"Become a little smaller after every update."**

---

# Why is it Called Weight Decay?

Suppose a weight is

```
10
```

After one update

```
9.9
```

Next update

```
9.8
```

Next update

```
9.7
```

The weights gradually **decay (shrink)** over time.

---

# Weight Update Without Weight Decay

$$
W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}
$$

---

# Weight Update With Weight Decay

$$
W_{new}=W_{old}-\eta\left(\frac{\partial L}{\partial W}+\lambda W_{old}\right)
$$

### Where,

- \(W_{old}\) = Current weight
- \(W_{new}\) = Updated weight
- \(\eta\) = Learning Rate
- \(\lambda\) = Weight Decay factor
- \(\frac{\partial L}{\partial W}\) = Gradient of the loss

Notice the additional term:

```
λW
```

This term continuously pulls the weights toward zero.

---

# How Weight Decay Works

Without Weight Decay

```
Weight

↓

12

↓

14

↓

18

↓

25
```

Weights keep increasing.

---

With Weight Decay

```
Weight

↓

12

↓

11.6

↓

11.2

↓

10.9
```

Weights remain controlled.

---

# Weight Decay vs L2 Regularization

Many beginners think these are exactly the same.

### In Classical Gradient Descent

They are mathematically equivalent.

Both encourage small weights.

---

### In Modern Optimizers (Adam, AdamW)

There is a difference.

- **L2 Regularization** adds the penalty directly to the loss function.
- **Weight Decay** shrinks the weights during the update step.

For optimizers like **AdamW**, Weight Decay is preferred because it decouples weight shrinkage from the adaptive gradient updates.

---

# Advantages of Weight Decay

- Reduces overfitting
- Keeps weights small
- Improves generalization
- Stable training
- Widely used with AdamW optimizer

---

# Disadvantages of Weight Decay

- Requires tuning the decay factor.
- Too much decay may cause underfitting.
- Can reduce learning if applied too aggressively.

---

# L1 vs L2 vs Weight Decay

| Feature | L1 | L2 | Weight Decay |
|----------|----|----|--------------|
| Penalty | Absolute weights | Squared weights | Shrinks weights during updates |
| Feature Selection | Yes | No | No |
| Sparse Model | Yes | No | No |
| Large Weights | Reduced | Reduced | Reduced |
| Small Weights | Become Zero | Become Smaller | Become Smaller |
| Common Use | Feature Selection | General Regularization | Deep Learning (especially AdamW) |

---

# Complete Workflow

```
Training Data
      ↓
Forward Propagation
      ↓
Prediction
      ↓
Loss Calculation
      ↓
Add Regularization Penalty
      ↓
Backpropagation
      ↓
Weight Update
      ↓
Smaller Weights
      ↓
Reduced Overfitting
      ↓
Better Generalization
```

---

# Key Points to Remember

- **Regularization** reduces **overfitting** by preventing the neural network from becoming too complex.
- **L2 Regularization (Ridge)** adds a penalty based on the **sum of squared weights**:

  $$
  Loss_{new}=Loss+\lambda\sum w_i^2
  $$

  It keeps weights **small** but does **not** make them zero.

- **L1 Regularization (Lasso)** adds a penalty based on the **sum of absolute weights**:

  $$
  Loss_{new}=Loss+\lambda\sum |w_i|
  $$

  It can make some weights **exactly zero**, thereby performing **automatic feature selection**.

- **Weight Decay** gradually shrinks weights during each optimization step. In simple Gradient Descent it is equivalent to L2 Regularization, but in modern optimizers like **AdamW**, it is implemented separately and generally provides better optimization behavior.
- Choosing the correct **regularization strength (λ)** is important because values that are too large can lead to **underfitting**.
